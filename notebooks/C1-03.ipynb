# Pythonでの複素数の扱い

```j```または```i```が虚数単位として使える。標準では```cmath```，勿論```numpy```でも使える。平方根や$\pi$をそのまま扱いたかったらSymPy。

## 標準
```complex```で実部，虚部を与えるのも可能。極形式で与えたい場合は```cmath.rect```または```cmath.exp```を使う。ただし角度単位はrad。

````python
import cmath
Z1 = complex(2,3)
Z2 = 3-2j
Z3 = 3.60*cmath.exp(1j*56.3*cmath.pi/180)
Z4 = cmath.rect(3.60, -33.7*cmath.pi/180)
display(Z1+Z2)
display(Z3/Z4)
````

極形式で表示したい場合は```cmath.polar```か，```abs```と```cmath.phase```を使う(前者は当然rad)

````python
Z5 = Z1+Z2
print(abs(Z5), cmath.phase(Z5)*180/cmath.pi)
display(cmath.polar(Z5))
````

## NumPy
標準の複素数をnpの関数群で受け取れるだけなのでcmathと大した違いはない。```np.angle```で```deg=True```オプションを付けるとdegreeで角度が得られるくらい。

````python
import numpy as np
Z1 = complex(2,3)
Z2 = 3-2j
Z3 = 3.60*np.exp(1j*56.3*np.pi/180)
Z4 = 3.60*np.exp(1j*-33.7*np.pi/180)
display(Z1+Z2)
display(Z3/Z4)
````

````python
Z5 = Z1+Z2
print(abs(Z5), np.angle(Z5,deg=True))
````

## SymPy
虚数単位は```sympy.I```。平方根とか$\pi$とかを綺麗に残して計算したい場合はこっちになる。望んだ出力にするのが案外面倒。

````python
from sympy import exp,I,pi, re, im, Abs, arg
Z1 = 2+3*I
Z2 = 3-2*I
Z3 = 3.60*exp(I*56.3*pi/180)
Z4 = 3.60*exp(I*-33.7*pi/180)
display(Z1+Z2)
display(Z3/Z4)
````

````python
Z5 = Z1+Z2
display(Z5)
display(Abs(Z5))
display(arg(Z5))
````

````python
re(Z5)+I*im(Z5)
````

````python
Z6 = Z3/Z4
display(Z6)
display(Abs(Z6))
display(arg(Z6))
````
