# SymPyでのインピーダンスやアドミタンスの計算
## $RLC$直列回路のインピーダンス
変数は```real=True```で定義しておくとsolveのときに実数として扱ってくれる。$j$でくくるには```expand_complex```

````python
from sympy import I, symbols, solve, Abs, arg, expand_complex, pi
R1, L1, C1, w = symbols('R_1 L_1 C_1 \omega', real=True)
ZR = R1
ZL = I*w*L1
ZC = 1/(I*w*C1)
Z = ZR+ZL+ZC
display(Z)
expand_complex(Z)
````

## $RLC$並列回路のアドミタンス

````python
G1 = 1/R1
BL = 1/ZL
BC = 1/ZC
Y = G1+BL+BC
display(Y)
expand_complex(Y)
````

## 値の代入

````python
Zs = Z.subs([(R1, 10), (L1, 10e-3), (C1, 100e-6), (w, 300)])
display(Zs)
display(Abs(Zs),arg(Zs)*180/float(pi))
````

````python
Ys = Y.subs([(R1, 10), (L1, 10e-3), (C1, 100e-6), (w, 300)])
display(Ys)
display(Abs(Ys),arg(Ys)*180/float(pi))
````

## Solveを使う
インピーダンスが$ 10+j8$で$L_1$が未知数の場合。式=0の形にしてsolveする

````python
solve(Z.subs([(R1, 10),(C1, 100e-6), (w, 300)])-(10+8j), L1)
````
