# SymPyでのインピーダンスやアドミタンスの計算
## 直並列回路のインピーダンス
複素数を代入するので```symbols```では実数指定はしない。$j$でくくるには```expand_complex```

````python
from sympy import I, symbols, solve, Abs, arg, expand_complex, pi
Z1, Z2, Z3, Z4 = symbols('\dot{Z}_1 \dot{Z}_2 \dot{Z}_3 \dot{Z}_4')
# Yp, Y2, Y3, Y4 = symbols('\dot{Y}_p \dot{Y}_2 \dot{Y}_3 \dot{Y}_4')
Y2 = 1/Z2
Y3 = 1/Z3
Y4 = 1/Z4
Yp = Y2+Y3+Y4
Zp = 1/Yp
Z = Z1+Zp
display(Z)
````

## 値の代入
subsで複素数を代入しても，そんなに綺麗な結果にはならない。```expand_complex```でも分数で出てくる。実数で出すには```evalf()```が必要

````python
Zv = Z.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5)])
display(Zv)
display(expand_complex(Zv))
display(expand_complex(Zv).evalf())
````

## 分圧，分流
式を作って代入するだけだが，```simplify```を使うと若干整理してくれる場合もある

````python
V = symbols('\dot{V}')
V1 = V*Z1/Z
display(V1)
V2 = V*Zp/Z
display(V2.simplify())
````

````python
V1.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf()
````

````python
V2.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf()
````

````python
IT = V/Z
display(IT)
IT.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf()
````

````python
I2 = IT*Y2/Yp
I3 = IT*Y3/Yp
I4 = IT*Y4/Yp
display(I2)
display(I3)
display(I4)
display(I2.simplify())
display(I3.simplify())
display(I4.simplify())

display(I2.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf())
display(I3.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf())
display(I4.subs([(Z1,7),(Z2,3*I),(Z3,-11*I),(Z4,5),(V,10)]).evalf())
````
