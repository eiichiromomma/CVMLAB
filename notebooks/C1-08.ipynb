# SymPyでの連立方程式の解法

```Matrix```で入れ子のリストを行列に変換する。```LDLsolve```で求解できる。

````python
from sympy import symbols, Matrix, I, MatMul, Eq
I1, I2 = symbols(r'\dot{I}_1 \dot{I}_2')
A = Matrix([
    [10+20*I, 20*I],
    [20*I, -10*I]])
B = Matrix([[100],[100*I]])
Iv = Matrix([[I1],[I2]])
print('係数行列')
display(A)
print('行列式')
display(A.det())
print('右辺')
display(B)
print('MatMulでevaluate=Falseとすれば式のまま')
leftside = MatMul(A,Iv,evaluate=False)
display(leftside)
print('Eqで式として定義できる')
display(Eq(leftside,B))
print('求解')
ans = A.LDLsolve(B).evalf()
display(ans)
display(Eq(Iv,ans))
# display(A.pinv_solve(B).evalf())
# display(A.cholesky_solve(B).evalf())
````

見た目がシンプルでも良ければNumPyで同様に解ける

````python
import numpy as np
A  = np.array([[10+20j, 20j],
              [20j, -10j]])
B = np.array([[100],[100j]])
np.linalg.solve(A,B)
````

## ガウス・ジョルダンの消去法
ガウス・ジョルダンの消去法では不確定な解がある場合は第二返り値として出力される

````python
import numpy as np
import sympy as sp
matrix = sp.Matrix(A)
b_matrix = sp.Matrix(B)
sol, freev = matrix.gauss_jordan_solve(b_matrix)
display(sol)
display(freev)
````

\[A B\]の状態にして```rref```も可能

````python
import numpy as np
from sympy import symbols, Matrix, I
M = [
    [10+20*I, 20*I, 100],
    [20*I, -10*I, 100*I]]
M = Matrix(M)
display(M)
rref_matrix, rref_pivots = M.rref()
display(rref_matrix)
display(rref_matrix[:,-1].evalf())
````
