# 共役を使った計算
SymPyでは```conjugate```が共役。

````python
from sympy import I, symbols, solve, Abs, arg, expand_complex, pi, conjugate, sqrt
R1, R2, R3, X1, X2 = symbols('R_1 R_2 R_3 X_1 X_2', real=True)
V = symbols('\dot{V}')
Z1 = R1+I*X1
Z2 = R2-I*X2
Z3 = R3
I1 = V/Z1
display(I1)
display(I1.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
I2 = V/Z2
display(I2)
display(I2.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
I3 = V/Z3
display(I3)
display(I3.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
````

````python
Yp = 1/Z1 + 1/Z2 + 1/Z3
display(Yp)
display(Yp.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
Zp = 1/Yp
display(Zp)
display(Zp.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
````

````python
P10=10*Abs(I1)**2
display(P10)
display(P10.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
P5 = 5*Abs(I2)**2
display(P5)
display(P5.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
P20 = 20*Abs(I3)**2
display(P20)
display(P20.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
````

複素電力を求める際に電圧の共役をとるため```conjugate(V)*IT```としている

````python
IT=Yp*V
display(IT.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
S = conjugate(V) * IT
display(S)
display(S.subs([(R1,10),(R2,5),(R3,20),(X1,5),(X2,10),(V,(10+10*I)/sqrt(2))]).evalf())
````
