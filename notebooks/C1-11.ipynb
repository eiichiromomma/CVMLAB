## 複素平面上での逆図形

```python
from sympy import I, symbols, conjugate, solve, im, re
from sympy.plotting import plot_implicit
import numpy as np
import matplotlib.pyplot as plt
al, be = symbols('alpha beta')
# a, bの複素平面上の2点を通る線 Adot を作る
a = 0.5J
b = 1+0.5J
ans = solve([al*re(a)+be-im(a), al*re(b)+be-im(b)], [al, be])
if len(ans) != 0:
    x = np.linspace(-5, 5, 10001) # 点数をリッチにしておかないと折れ線になる
    y = (ans[al]*x + ans[be]).astype(np.float32)
else:
    x = np.repeat(float(re(a)),10001)
    y = np.linspace(-5,5,10001)
Adot= x+1J*y
if 0 in Adot:
    Adot[Adot==0] = 0.000001
InvAdot=1/Adot
x2 = InvAdot.real
y2 = InvAdot.imag
fig, ax = plt.subplots()
ax.plot(x,y,'r-')
ax.plot(x2,y2,'-',label=r'$\frac{1}{\dot{A}}$') #ax1グラフを描く
ax.set_xlim(-5,5)
ax.set_ylim(-5,5)
ax.set_xlabel('real',loc='right')
ax.set_ylabel('imag',loc='top')
ax.set_aspect('equal')
ax.spines["right"].set_color("none")      # グラフ右側の軸(枠)を消す
ax.spines["top"].set_color("none")        # グラフ上側の軸(枠)を消す
ax.spines['left'].set_position('zero')    #グラフ左側の軸(枠)が原点を通る
ax.spines['bottom'].set_position('zero')  #グラフ下側の軸(枠)が原点を通る

```
